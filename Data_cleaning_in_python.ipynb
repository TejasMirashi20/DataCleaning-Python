```python
import pandas as pd
df = pd.read_excel("/content/messy_customer_dataset(1).xlsx")
df
```

```python
df=df.drop_duplicates()
```

```python
df['Last_Name'] = df['Last_Name'].str.strip("123./_")
```

```python
df
```

```python
df['Phone_Number'] = df['Phone_Number'].str.replace('[^a-zA-Z0-9]', '', regex=True)
```

```python
df
```

```python
df['Phone_Number'] = df['Phone_Number'].apply(lambda x: str(x))
```

```python
df
```

```python
df['Phone_Number'] = df['Phone_Number'].str.replace('[^0-9]', '', regex=True)
```

```python
df['Phone_Number'] = df['Phone_Number'].apply(lambda x: x[0:3] + '-' + x[3:6] + '-' + x[6:10])
```

```python
df
```

```python
df[['Street_Name', 'Street_Type', 'Street_Number']] = df['Address'].str.split(',', expand=True)
```

```python
df
```

```python
df= df.fillna('')
df
```

```python
df = df.rename(columns={
    'Street_Name': 'Street',
    'Street_Number': 'Type',
    'Street_Type': 'Number'
})
```

```python
df
```

```python
df = df.drop(columns=['Not_Useful_Column'])
```

```python
df['Do_Not_Contact'] = df['Do_Not_Contact'].str.replace('Yes', 'Y')
df['Do_Not_Contact'] = df['Do_Not_Contact'].str.replace('No', 'N')
df
```

```python
for x in df.index:
  if df.loc[x,"Do_Not_Contact"] == 'Y':
    df.drop(x, inplace=True)
df
```

```python
for x in df.index:
  if df.loc[x,"Do_Not_Contact"] == '':
    df.drop(x, inplace=True)
df
```

```python
for x in df.index:
  if df.loc[x,"Phone_Number"] == '--':
    df.drop(x, inplace=True)
df
```

```python
df = df.reset_index(drop=True)
df
```

```python
df = df.drop(columns=['Number'])
```
